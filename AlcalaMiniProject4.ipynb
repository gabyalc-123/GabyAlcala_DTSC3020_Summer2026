## Setup — Run This First

This cell imports the required libraries and defines `get_soup(url)`.

The helper tries the live website first. If the website cannot be reached, it uses a small backup page with the same HTML structure.

**Do not change this setup cell.**

````python
import requests
from bs4 import BeautifulSoup
import csv
import os
import pandas as pd

# A small backup page with the same structure as Quotes to Scrape.
BACKUP_HTML = """
<html>
<body>
<div class="quote">
    <span class="text">"The world as we have created it is a process of our thinking."</span>
    <small class="author">Albert Einstein</small>
    <div class="tags">
        <a class="tag">change</a>
        <a class="tag">thinking</a>
    </div>
</div>

<div class="quote">
    <span class="text">"It is our choices that show what we truly are."</span>
    <small class="author">J.K. Rowling</small>
    <div class="tags">
        <a class="tag">choices</a>
        <a class="tag">abilities</a>
    </div>
</div>

<div class="quote">
    <span class="text">"Try not to become a person of success, but a person of value."</span>
    <small class="author">Albert Einstein</small>
    <div class="tags">
        <a class="tag">success</a>
        <a class="tag">value</a>
    </div>
</div>
</body>
</html>
"""

def get_soup(url):
    """Download a webpage and return a BeautifulSoup object."""
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        return BeautifulSoup(response.text, "html.parser")
    except requests.RequestException:
        print("Live website unavailable — using the backup page.")
        return BeautifulSoup(BACKUP_HTML, "html.parser")

print("Setup complete. get_soup(url) is ready.")
````

# Part 2 — Extract One Quote  
## 20 points

Use the `soup` object from Part 1.

### Requirements

1. Use `find("div", class_="quote")` to find the first quote block.
2. Extract and print:
   - the quote text
   - the author
3. Use `.get_text(strip=True)` to remove extra spaces.

Your output should follow this format:

```text
Text: ...
Author: ...
```

### Reminder

- `find(...)` returns the first matching element.
- `class_` has an underscore because `class` is a reserved Python word.

````python
# Name: Gaby Alcala
# Part 2 — Extract one quote

# Find the first quote block
first_quote = soup.find('div', class_='quote')

# Find the <span class='text'>
text_element = first_quote.find('span', class_='text')

# Find the <small class='author'>
author_element = first_quote.find('small', class_='author')

# Extract text and remove visible spaces
text = text_element.get_text(strip=True)
author = author_element.get_text(strip=True)

# Print results
print('Text:', text)
print('Author:', author)
````

# Part 3 — Extract All Quotes  
## 30 points

Now collect every quote on the page.

### Requirements

1. Use `find_all("div", class_="quote")` to collect all quote blocks.
2. Print the number of quotes found.
3. Create an empty list named `records`.
4. Loop through all quote blocks.
5. For each quote, extract:
   - `text`
   - `author`
   - `tags`
6. Store each quote as a dictionary:

```text
{"author": author, "text": text, "tags": tags}
```

7. Append each dictionary to `records`.
8. Print each author and their list of tags.

### Expected structure

```text
Author Name -> ['tag1', 'tag2']
```

### Reminder

- `find(...)` returns one match.
- `find_all(...)` returns a list of all matches.
- The `tags` value should still be a Python list in this part.

````python
# Name: Gaby Alcala
# Part 3 — Extract all quotes

# Find every <div class = 'quote'>
quotes = soup.find_all('div', class_='quote')

# Print the number of quotes found
print('Quotes found:', len(quotes))

# Create empty list
records = []

# Loop through all quote blocks and extract text, author, and tags
for quote in quotes:
    text = quote.find(    # Extract the quote text
        'span',
        class_='text'
    ).get_text(strip=True)

    author = quote.find(     # Extract the author
        'small',
        class_='author'
    ).get_text(strip=True)

    tag_elements = quote.find_all(  # Find all tags using find_all
        'a',
        class_='tag'
    )

    # Store tags in Python list
    tags = []

    for tag_element in tag_elements:
        tag_text = tag_element.get_text(strip=True)
        tags.append(tag_text)

    # Create one structured record per quote
    record = {
        'author': author,
        'text' : text,
        'tags' : tags
    }

    # Append each dictionary to records
    records.append(record)

for record in records:
    print(record['author'], '->', record['tags'])
````

Use the `records` list from Part 3.

````python
# Name: Gaby Alcala
# Part 4 — Save and analyze the data

with open('quotes.csv', 'w', newline='',encoding='utf-8') as file:

    # DictWriter to write dict as CSV rows with fieldnames(column names)
    writer = csv.DictWriter(file, fieldnames=['author', 'text', 'tags'])

    # Write header names using first row
    writer.writeheader()

    # Loop through records to write one row per record
    for record in records:
        writer.writerow({
            'author': record['author'],
            'text': record['text'],

            # Convert tag list into one string
            'tags': ';'.join(record['tags'])
        })

print('CSV exists?', os.path.exists('quotes.csv'))

# Load csv file into pandas DataFrame named df
df = pd.read_csv('quotes.csv')

# Display the first five rows using head()
print('\nFirst five rows:')
print(df.head())

# Print DataFrame shape (# of rows and columns)
print('\nShape:', df.shape)

# Print # of quotes per author
print('\nQuotes per author:')
print(df['author'].value_counts())

# find() returns the first match, find_all() returns all the matches
# One responsible web-scraping rule: Do not bypass login or CAPTCHA
````
